```python
import pandas as pd
import  matplotlib.pyplot as plt
```

```python
#path to the csv file
csv_path="fraud_dataset.csv"
```

```python
#Readinng/Creating the DataFrame from the csv file.
df=pd.read_csv(csv_path)
```

```python
#Checking overall columns data.
df.info()
```

```python
#Checking the top 5 rows
df.head()
```

```python
#Checking sum of null values in each column
df.isna().sum()
```

```python
#Checking how the data is distributed in all numerical columns
df.describe()
```

```python
#Fixing the date and time format.
df['transaction_time'] = pd.to_datetime(df['transaction_time'], format='mixed', errors='coerce')
```

```python
#handling missing values.
df['customer_age']=df['customer_age'].fillna(df['customer_age'].median())

#Replaced missing values in category and merchant columns with "Unknown" str 
#cause there were 500 missing values in both columns 
#with the mode in both been closest to the second largest value 
#so giving it the value "Unknown" seems better than some very specific values and creating bias.
df['category']=df['category'].fillna("Unknown")
df['merchant']=df['merchant'].fillna("Unknown")
```

```python
#Checking if there are any missing values left even after replacing.
df.isna().sum()
```

```python
#Dectecting outliers using basic Interquartile Range Method in the amount column.
Amount_Q1 = df['amount'].quantile(0.25)
Amount_Q3 = df['amount'].quantile(0.75)
Amount_IQR = Amount_Q3 - Amount_Q1
Amount_lower = Amount_Q1 - 1.5 * Amount_IQR
Amount_upper = Amount_Q3 + 1.5 * Amount_IQR
Amount_outliers = df[
    (df['amount'] < Amount_lower) |
    (df['amount'] > Amount_upper)
]
```

```python
#Printing Amount Column Outliers.
print(Amount_outliers)
print(Amount_outliers.shape[0])
```

```python
Amount_outliers.loc[Amount_outliers['is_fraud']==1].shape[0]
```

```python
#Dectecting outliers using basic Interquartile Range Method in the avg_spend column.
Avg_Spend_Q1 = df['avg_spend'].quantile(0.25)
Avg_Spend_Q3 = df['avg_spend'].quantile(0.75)
Avg_Spend_IQR = Avg_Spend_Q3 - Avg_Spend_Q1
Avg_Spend_lower = Avg_Spend_Q1 - 1.5 * Avg_Spend_IQR
Avg_Spend_upper = Avg_Spend_Q3 + 1.5 * Avg_Spend_IQR
Avg_Spend_outliers = df[
    (df['avg_spend'] < Avg_Spend_lower) |
    (df['avg_spend'] > Avg_Spend_upper)
]
```

```python
#Printing Avg_Spend Column Outliers.
print(Avg_Spend_outliers)
print(Avg_Spend_outliers.shape[0])
```
